```python
def eliminar_columnas(df):
    """Elimina de df_article las columnas marcadas como innecesarias.
    Devuelve un DataFrame nuevo (no modifica el original)."""

    cols_sobran = [
        'CodCompany','Notes','Image','LocationWarehouseControl','SeriesControl',
        'UnitarianSeries','ExpiryControl','ExpireDays','IDUnitQuantityWarehouse', 'IDUnitPriceWarehouse',
        'SecondUnit','IDUnitSecondUnit','IDUnitQuantityPurchase','IDUnitPricePurchase','IDProductSubFamily',
        'StatisticsCode','CodCountryOfOrigin','SupplemetUnit','IDUnitSupplement','AverageSecondUnit',
        'Dimension','ApplyPurchaseDiscount','ApplyPurFinancialSurcharge','IDAttribute1',
        'IDAttribute2','IDAttribute3','IDAttributeValue1','IDAttributeValue2','IDAttributeValue3',
        'IDArticleGeneric','CodArticlePrevious','IDUnitHomogeneousPurchase','PlanningCategory', 'PlanningGroup',
        'PlanningGroup','NetWeight','MakeAutomaticOutput','ImputationCostUpdated', 'InvestmentExpense',
        'IDUnitQuantitySales','IDUnitPriceSales','IDUnitPriceSales','ApplySalesDiscount',
        'ApplySalesFinancialSurcharge','IDUnitHomogeneousSales','AverageSecondUnitSerie','GrossWeight', 'Container',
        'HasContainer','DiscountContainer','IDStatusLotGroup','StatusLot','ConsumePrevisionBehavior',
        'PDMItemID','ImputationTotalQuantity','GroupDaysNumber','TechnicalDescription','MRPPrevisionsOnly',
        'ApplySalesRetention','ApplyPurchaseRetention','RowTimestamp','LastModifiedBy',
        'QuantityBaseFormula', 'IDUnitBaseFormula', 'PotencyControl', 'PotencyDefault', 'BatchBalancing', 'MultipleCADLink',
        'ComponentsGroupType', 'InactiveDate', 'WholeSeries', 'LotsReserve', 'LotsReserveDays', 'LotsOrder'
        'LotsOrderDays', 'SecondUnitFixed', 'MatchCode', 'UnitarianSerieOrder', 'IDArticleContainer', 'CreationTimestamp',
        'RowCreatedBy', 'ApplyCommission', 'ApplySalesDiscountPP', 'ApplyPurchaseDiscountPP', 'IDArticleClassification', 
        'ManufacturingGroupCode', 'ManufacturingGroupQuantity', 'MaterialPreparationType', 'NotBillablePurchase', 'NotBillableSales',
        'NotApplyNetWeightNetMass', 'UDI', 'UDIBasic', 'HSCode', 'GroupDays', 'LotsOrder', 'LotsOrderDays'
    ]

    cols_a_quitar = [c for c in cols_sobran if c in columns_article]
    df_out = df.drop(columns=cols_a_quitar)

    # Aviso por si alguna de la lista no se encontró (typo o nombre distinto)
    no_encontradas = [c for c in cols_sobran if c not in df_article.columns]
    if no_encontradas:
        print(f"Aviso: {len(no_encontradas)} columnas de la lista no estaban en df: {no_encontradas}")

    print(f"Eliminadas {len(cols_a_quitar)} columnas. "
          f"Quedan {df_out.shape[1]} columnas, {df_out.shape[0]} filas.")
    return df_out
```

```python
df_article = eliminar_columnas(df_article)
```

```python
df_article = df_article.rename(columns={
    'CodArticle': 'Código Artículo',
    'Description': 'Descripción Artículo',
    'IDProductLine': 'Línea de Producto',
    'IDProductFamily': 'Sublínea de Producto'
})

display(df_article.sample(2))
```

## Mapear los id's con nombres

```python
ruta_line = os.path.join("res\\raw", "ProductLine.csv")
print(ruta_line) 
ruta_family = os.path.join("res\\raw", "ProductFamily.csv")
print(ruta_family)

columns_line = ['IDProductLine',	'CodCompany',	'CodProductLine',	'Description',	'Image',	'RowTimestamp',	'LastModifiedBy',	'CreationTimestamp',	'RowCreatedBy']
columns_family = ['IDProductFamily',	'CodCompany',	'IDProductLine',	'CodProductFamily',	'Description',	'Image',	'RowTimestamp',	'LastModifiedBy',	'CreationTimestamp',	'RowCreatedBy']

pd.set_option("display.max_columns", None)  
pd.set_option("display.width", None)               

df_line = pd.read_csv(ruta_line, sep=";", header=None, names=columns_line)
df_family = pd.read_csv(ruta_family, sep=";", header=None, names=columns_family)

```

```python
display(df_line.sample(1))
display(df_line.shape)
display(df_family.sample(1))
display(df_family.shape)
```

```python
df_article = df_article.merge(df_line[['IDProductLine', 'CodProductLine', 'Description']].rename(columns={'Description': 'Nombre Línea'}), 
                              left_on="Línea de Producto", 
                              right_on="IDProductLine", 
                              how="left")
```

```python
df_article = df_article.merge(df_family[['IDProductFamily', 'CodProductFamily', 'Description']].rename(columns={'Description': 'Nombre Família'}), 
                              left_on="Sublínea de Producto", 
                              right_on="IDProductFamily", 
                              how="left")
```

```python
display(df_article.sample(1))
```

```python
df_article = df_article.drop(columns=['IDArticle', 'Línea de Producto', 'Sublínea de Producto', 'IDProductLine', 'IDProductFamily'])
display(df_article)
```

## Cargamos las imputaciones

```python
df_FASE_final = pd.read_parquet("res//procesed/df_FASE_final.parquet")

pd.set_option("display.max_columns", None)
pd.set_option("display.width", None)   

display(df_FASE_final.head(1))
```

```python
df_FASE_final = df_FASE_final.merge(df_article, 
                                    left_on="smMaterialId", 
                                    right_on="Código Artículo", 
                                    how="left")

display(df_FASE_final[df_FASE_final['smOrdenId']=='OF2509545'])
display(df_FASE_final.shape)
```

```python
df_final = df_FASE_final 
```

```python
print(f'Número de Líneas de Productos distitnas: {len(df_final['CodProductLine'].unique())}')
print(f'Número de Famílias de Productos distitnas: {len(df_final['CodProductFamily'].unique())}')
print(f'')

print(f'Número de fases sin Linea de Producto: {df_final['CodProductLine'].isnull().sum()}')
```

```python
#df_final.loc[df_final['CodProductLine'].isnull(), 'CodProductLine'] = 'Sin Linea'
#df_final.loc[df_final['Nombre Família'].isnull(), 'Nombre Família'] = 'Sin Família'
```

```python
conteo = df_final.groupby(["Nombre Línea", "Nombre Família"]).size().reset_index(name="conteo")
conteo = conteo.sort_values(["Nombre Línea", "conteo"], ascending=[True, True])

# Crear etiqueta combinada para mostrar la jerarquía
conteo["etiqueta"] = conteo["Nombre Línea"] + " → " + conteo["Nombre Família"]

lineas_unicas = conteo["Nombre Línea"].unique()
colores = cm.tab20(range(len(lineas_unicas)))
color_map = dict(zip(lineas_unicas, colores))

fig, ax = plt.subplots(figsize=(12, len(conteo)*0.3))
bar_colors = conteo["Nombre Línea"].map(color_map)
ax.barh(conteo["etiqueta"], conteo["conteo"], color=bar_colors)
ax.set_xscale("log")
ax.set_xlabel("Nº de fases (escala log)")
plt.title("Distribución de fases por línea y familia")
plt.tight_layout()
plt.show()
```
