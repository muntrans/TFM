# Análisis de viabilidad del TFM
En `df_fases_MES`tengo la **fase**, **máquina dominante**, y el **grupo/familia de esa máquina**. Con eso puedo montar un  cruce fase × máquina dentro de cada familia, ahi es donde vive la decisión de asignación.

Entonces, la elección real no es "esta fase fue a esta máquina" (eso siempre pasa). La elección existe cuando fases del mismo tipo pudieron ir a máquinas distintas de la misma familia.   
Por eso el análisis interesante es dentro de cada grupo: `¿las fases de una familia se reparten entre sus varias máquinas, o cada tipo de fase va siempre a la misma?`

Cuántas máquinas y cuántas fases tiene cada familia:

```python
resumen_familia = df_final.groupby("NombreGrupo").agg(
    n_fases=("SmFaseUid", "nunique"),
    n_maquinas=("MaquinaDominante", "nunique"),
).sort_values("n_fases", ascending=False)

display(resumen_familia)
print(f"Total grupos de máquinas: {len(resumen_familia)}")
```

Toca observar si las imputaciones se reparten entre las máquinas de la família.

```python
familias_analizables = set(df_final["NombreGrupo"])

for familia in familias_analizables:
    sub = df_final[df_final["NombreGrupo"] == familia]
    reparto = sub["MaquinaDominante"].value_counts(normalize=True)
    print(f"{familia[:45]:45s} | fases: {len(sub):5d} | máq. principal: {reparto.iloc[0]:.0%} | nº máq. usadas: {(reparto > 0.001).sum()}")
```

```python
familias_analizables = set(df_final["NombreGrupo"])

resultado = []
for familia in familias_analizables:
    subset = df_final[df_final["NombreGrupo"] == familia]
    n_fases_distintas = subset["Description"].nunique()
    resultado.append({"Nombre grupo": familia, "Cantidad tipos fases": n_fases_distintas})

contando_fases = pd.DataFrame(resultado)
display(contando_fases)
```

Vamos a trabajar con tiempo total, en las imputaciones aveces todo el peso se queda en preparación o producción, sin separlo, esto se debe a una mala disciplina entre los operarios. 

```python
# ¿cuántas fases tienen teórico en una parte pero real 0 en esa misma parte?
prep_teorico_sin_real = ((df_final["PreparationTime"] > 0) & (df_final["MinutosPreparacion_efectivo"] == 0)).sum()
prod_teorico_sin_real = ((df_final["ExecutionTime"] > 0) & (df_final["MinutosProduccion_efectivo"] == 0)).sum()

print(f"Fases con teórico de preparación pero real de preparación = 0: {prep_teorico_sin_real}")
print(f"Fases con teórico de producción pero real de producción = 0: {prod_teorico_sin_real}")
print(f"Total fases: {len(df_final)}\n")
```

Casi todas las fases tienen teórico en una parte pero real 0 en esa misma parte, significa que la separación preparación/producción en el MES no es fiable: el trabajo se imputa a una casilla u otra sin correspondencia con cómo lo definió el ERP.

```python
# Inspeccionar las maquinas de cada familia:
maquinas_por_familia = df_final.groupby(["NombreGrupo", "MaquinaDominante"]).size()

for familia, maquinas in maquinas_por_familia.groupby(level=0):
    print(f"{familia}:")
    print(maquinas.droplevel(0).sort_values(ascending=False).to_string())
    print()
```

## Decisión de Viabilidad 
Tras el proceso de limpieza y depuración realizado, se dispone de un bloque sólido de familias: 9 familias con elección real y volumen suficiente.

Este resultado cierra el análisis de viabilidad, con una conclusión positiva.

- Torns Control Numèric — 34% de concentración en la máquina principal, 4 máquinas utilizadas. El trabajo se reparte de forma equilibrada entre las cuatro, lo que valida la fusión de grupos realizada y constituye el ejemplo más claro de elección real. 1.641 fases.
- Centres Vertical COURES — 49% de concentración, 4 máquinas utilizadas. Reparto saludable sobre 2.636 fases.
- Premses Conformats — 56% de concentración, 2 máquinas utilizadas. Reparto equilibrado entre las dos máquinas activas, con un volumen moderado (314 fases).
- Centres Vertical ALUMINIS — 65% de concentración, 3 máquinas utilizadas. Reparto adecuado sobre 507 fases.
- Centres Horitzontal COURES — 73% de concentración, 3 máquinas utilizadas. Reparto aceptable sobre 1.242 fases.

Estas cinco familias constituyen el núcleo sólido del análisis: presentan una decisión de asignación efectiva y un reparto real del trabajo entre máquinas.

- Torns Puma — 79% de concentración, 2 máquinas utilizadas. Aunque la máquina principal concentra la mayor parte del trabajo, el 21% restante sobre 4.088 fases representa un volumen relevante en la segunda máquina.
- Centre 5 Eixos petites — 80% de concentración, 4 máquinas utilizadas. La máquina principal domina, pero existe reparto entre las cuatro máquinas. 553 fases.
- Màquines de Taladrat Profund — 89% de concentración, 2 máquinas utilizadas. Una máquina concentra casi todo el trabajo, aunque el 11% restante sobre 3.604 fases sigue representando un volumen relevante en la segunda.

Estas tres familias se consideran frontera: se incluyen en el análisis, si bien podrían quedar excluidas según evolucione el estudio.

- Màquines Secció Flexibles — 98% de concentración, y solo 174 fases. La máquina principal concentra casi la totalidad del trabajo sobre un volumen bajo, por lo que la elección resulta prácticamente inexistente. Se propone excluir esta familia del análisis de idoneidad, o mencionarla explícitamente como familia sin elección efectiva.

## Conclusión
De las familias productivas con múltiples máquinas, la mayoría presentan un reparto efectivo del trabajo entre sus máquinas, lo que confirma que la asignación fase-máquina constituye una decisión real y no determinista en el taller.

## Justificación
Análisis de viabilidad: existencia de una decisión de asignación real

Antes de abordar el modelado, resulta imprescindible verificar una premisa sobre la que se sostiene todo el trabajo: que la asignación de una fase a una máquina constituye efectivamente una decisión con alternativas, y no un proceso determinista en el que cada fase se ejecuta necesariamente en un único recurso posible. Si no existiera tal elección, no habría idoneidad que evaluar ni asignación que optimizar, y el problema carecería de sentido.

Para comprobarlo, se ha analizado el conjunto de familias de máquina productivas resultantes del proceso de depuración, cada una de las cuales agrupa máquinas funcionalmente equivalentes entre las que el taller puede repartir el trabajo. Se ha exigido, como condición necesaria, que una familia disponga de al menos dos máquinas; las familias constituidas por una única máquina se excluyeron de este análisis por no ofrecer alternativa alguna de asignación. Sobre las familias restantes, se ha examinado no solo cuántas máquinas podrían ejecutar sus fases, sino en qué medida el trabajo se reparte realmente entre ellas, calculando la proporción de fases que concentra la máquina más utilizada de cada familia.

Los resultados confirman la existencia de una decisión de asignación efectiva. En la mayoría de las familias analizadas, la máquina principal concentra entre el 34 % y el 73 % de las fases, evidenciando que el trabajo se distribuye de forma genuina entre varios recursos y que, por tanto, existe una elección real en la asignación. Casos como el de los tornos de control numérico (34 % en la máquina principal, con las fases repartidas entre cuatro máquinas) o los centros de mecanizado vertical de cobre (49 %) constituyen la evidencia más clara de que la asignación no es determinista. Se observaron asimismo familias con una concentración elevada en una única máquina (superior al 85 %), que, si bien disponen de alternativas, presentan una elección efectiva más limitada; estas se tratan con cautela en el análisis posterior.

En conjunto, este análisis valida empíricamente la premisa central del trabajo: existe un problema de asignación fase-máquina con capacidad de elección real, sobre un volumen de datos suficiente, que un modelo de idoneidad puede aprender a caracterizar y optimizar.

En conjunto, tras el proceso completo de depuración, el análisis se sostiene sobre un total de 14.759 fases distribuidas en las familias de máquina productivas con capacidad de elección, un volumen que proporciona una base de datos amplia y suficiente para el modelado posterior.
